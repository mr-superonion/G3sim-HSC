```python
%pylab inline
import gc
import os
import glob
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
import healpy as hp
import astropy.io.fits as pyfits
import astropy.table as astTab
from configparser import ConfigParser
import sim_analysis_utilities as utilities
```

```python
configName  =   'wrkDir/s19/s19-1/anaCat_newS19Mask_fdeltacut/plot/char.ini'
parser      =   ConfigParser()
parser.read(configName)
```

```python
dataDir='wrkDir/s19/s19-1/anaCat_newS19Mask_fdeltacut/catalog_others/'
numRan=0.
for fieldname in utilities.field_names:
    if not parser.has_section(fieldname):
        parser.add_section(fieldname)
    data=pyfits.getdata(os.path.join(dataDir,'%s_rand.fits' %fieldname))
    n=len(data)
    parser.set(fieldname,'area-deg2','%.2f' %(n/100./3600.))
    numRan+=n
    print(n)
    del data
    gc.collect()
if not parser.has_section('All'):
    parser.add_section('All')
parser.set('All','area-deg2','%.2f' %(numRan/100./3600.))
```

```python
inDir       =   '/home/xiangchong/work/gw_ipmu/work/massMapSparse/s19/reGaus_catalog/'
numGal=0.
wsAll=0.
for fieldname in utilities.field_names:
    if not parser.has_section(fieldname):
        parser.add_section(fieldname)
    data=pyfits.getdata(os.path.join(inDir,'%s.fits' %fieldname))
    ng=len(data)
    ws=np.sum(data['weight'])
    parser.set(fieldname,'ws-ave','%.2f' %(ws/ng))
    parser.set(fieldname,'n-arcmin2','%.2f' %(ng/parser.getfloat(fieldname,'area-deg2')/3600.))
    numGal+=ng
    wsAll+=ws
if not parser.has_section('All'):
    parser.add_section('All')
parser.set('All','ws-ave','%.2f' %(wsAll/numGal))
parser.set('All','n-arcmin2','%.2f' %(numGal/parser.getfloat('All','area-deg2')/3600.))
```

```python
with open(configName, 'w') as configfile:
    parser.write(configfile)
```
