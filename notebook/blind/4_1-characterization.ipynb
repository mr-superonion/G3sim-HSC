```python
configName  =   'wrkDir/s19/s19-1/anaCat_newS19Mask_fdeltacut/plot/char.ini'
parser      =   ConfigParser()
parser.read(configName)
```

```python
#Area from random catalog
dataDir='wrkDir/s19/s19-1/anaCat_newS19Mask_fdeltacut/catalog_others/'
numRan=0.
for fieldname in utilities.field_names:
    if not parser.has_section(fieldname):
        parser.add_section(fieldname)
    data=pyfits.getdata(os.path.join(dataDir,'%s_rand.fits' %fieldname))
    n=len(data)
    parser.set(fieldname,'area-deg2','%.2f' %(n/100./3600.))
    numRan+=n
    print(n)
    del data
    gc.collect()
if not parser.has_section('All'):
    parser.add_section('All')
parser.set('All','area-deg2','%.2f' %(numRan/100./3600.))
```

```python
#Seeing
dataDir='wrkDir/s19/s19-1/anaCat_newS19Mask_fdeltacut/catalog_obs_reGaus_no_m/'
fwhmAll=0.
numAll=0.
neffAll=0.
ermsAll=0.
sigmaeAll=0.
for fieldname in utilities.field_names:
    if not parser.has_section(fieldname):
        parser.add_section(fieldname)
    data=pyfits.getdata(os.path.join(dataDir,'%s_no_m.fits' %fieldname))
    n=len(data)
    neff=np.sum(data['i_hsmshaperegauss_derived_rms_e']**2./(data['i_hsmshaperegauss_derived_rms_e']**2.+data['i_hsmshaperegauss_derived_sigma_e']**2.))
    parser.set(fieldname,'neff-arcmin2','%.2f' %(neff/parser.getfloat(fieldname,'area-deg2')/3600.))
    fwhm=np.sum(utilities.get_psf_size(data))
    erms=np.sum(data['i_hsmshaperegauss_derived_rms_e'])
    sigmae=np.sum(data['i_hsmshaperegauss_derived_sigma_e'])
    parser.set(fieldname,'seeing-ave','%.2f' %(fwhm/n))
    parser.set(fieldname,'erms-ave','%.2f' %(erms/n))
    parser.set(fieldname,'sigmae-ave','%.2f' %(sigmae/n))
    print(fwhm/n)
    print(erms/n)
    print(sigmae/n)
    numAll+=n
    neffAll+=neff
    fwhmAll+=fwhm
    ermsAll+=erms
    sigmaeAll+=sigmae
    print(np.average(data['i_hsmshaperegauss_derived_sigma_e']**2.+data['i_hsmshaperegauss_derived_rms_e']**2.)*2.)
    print(np.average(data['i_hsmshaperegauss_e2']**2.+data['i_hsmshaperegauss_e1']**2.))
    del data
    gc.collect()
if not parser.has_section('All'):
    parser.add_section('All')
parser.set('All','seeing-ave','%.2f' %(fwhmAll/numAll))
parser.set('All','erms-ave','%.2f' %(ermsAll/numAll))
parser.set('All','sigmae-ave','%.2f' %(sigmaeAll/numAll))
parser.set('All','neff-arcmin2','%.2f' %(neffAll/parser.getfloat('All','area-deg2')/3600.))
```

```python
# weight,e_rms,sigma_e,n,neff
inDir       =   '/home/xiangchong/work/gw_ipmu/work/massMapSparse/s19/reGaus_catalog/'
numGal=0.
wsAll=0.
for fieldname in utilities.field_names:
    if not parser.has_section(fieldname):
        parser.add_section(fieldname)
    data=pyfits.getdata(os.path.join(inDir,'%s.fits' %fieldname))
    ng=len(data)
    ws=np.sum(data['weight'])
    parser.set(fieldname,'ws-ave','%.2f' %(ws/ng))
    parser.set(fieldname,'n-arcmin2','%.2f' %(ng/parser.getfloat(fieldname,'area-deg2')/3600.))
    numGal+=ng
    wsAll+=ws
if not parser.has_section('All'):
    parser.add_section('All')
parser.set('All','ws-ave','%.2f' %(wsAll/numGal))
parser.set('All','n-arcmin2','%.2f' %(numGal/parser.getfloat('All','area-deg2')/3600.))
```
