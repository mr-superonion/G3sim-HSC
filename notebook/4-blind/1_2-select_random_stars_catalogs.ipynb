```python
%pylab inline
import gc
import os
import glob
import numpy as np

import matplotlib
import matplotlib.pyplot as plt
import healpy as hp

import astropy.io.fits as pyfits

import sim_analysis_utilities as utilities

def get_FDFC_flag(ra,dec,version='s19',degree=True):
    # Read fdfc file
    datDir='/gpfs02/work/xiangchong.li/S19ACatalogs/'
    hpfname =   os.path.join(datDir,"s19a_fdfc_hp_contarea_izy-gt-5_trimmed_fd001.fits") 
    m       =   hp.read_map(hpfname, nest = True, dtype = np.bool)

    # Get flag
    if degree:
        mfactor =   np.pi/180.
    else:
        mfactor =   1.
    indices_map =   np.where(m)[0]
    nside   =   hp.get_nside(m)
    phi     =   ra*mfactor
    theta   =   np.pi/2. - dec*mfactor
    indices_obj = hp.ang2pix(nside, theta, phi, nest = True)
    return np.in1d(indices_obj, indices_map)
```

```python
dataDir='/gpfs02/work/miyatake/S19A/S19A_star_mask_g15_morecut/'
outDir='/gpfs02/work/xiangchong.li/S19ACatalogs/v2/catalog_others/'
if not os.path.isdir(outDir):
    os.mkdir(outDir)
for fieldname in utilities.field_names:
    oname=os.path.join(outDir,'%s_rand.fits' %fieldname)
    if os.path.isfile(oname):
        continue
    data=pyfits.getdata(os.path.join(dataDir,'%s_rand.fits' %fieldname))
    print(len(data))
    mask=get_FDFC_flag(data['ra'],data['dec'])
    if fieldname=='GAMA09H':
        print(fieldname)
        print(len(data))
        mask&=utilities.get_mask_visit_104994(data)
    data=data[mask]
    print(len(data))
    pyfits.writeto(oname,data,overwrite=True)
    del data
    del mask
    gc.collect()
```
