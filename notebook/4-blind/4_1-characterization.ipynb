# Total Area

```python
#Area from random catalog
dataDir='wrkDir/s19/s19-1/anaCat_newS19Mask_fdeltacut/catalog_others/'
numRan=0.
for fieldname in utilities.field_names:
    if not parser.has_section(fieldname):
        parser.add_section(fieldname)
    data=pyfits.getdata(os.path.join(dataDir,'%s_rand.fits' %fieldname))
    n=len(data)
    parser.set(fieldname,'area-deg2','%.2f' %(n/100./3600.))
    numRan+=n
    print(n)
    del data
    gc.collect()
if not parser.has_section('All'):
    parser.add_section('All')
parser.set('All','area-deg2','%.2f' %(numRan/100./3600.))
```
