```python
%pylab inline
import os
import numpy as np

import pandas as pd
import matplotlib
import astropy.io.fits as pyfits
import astropy.io.ascii as ascii
import matplotlib.pyplot as plt

import sim_analysis_utilities as utilities
```

```python
anaDir='./anaCat_newS19Mask_fdeltacut/'
inFname=os.path.join(anaDir,'allSim.h5')
allSim=pd.read_hdf(inFname,key='data')
info=ascii.read('info_newS19Mask_fd001_sim.csv')
```

```python
for fieldname in utilities.field_names:
    outDir=os.path.join(anaDir,fieldname,'smallcats')
    if not os.path.exists(outDir):
        os.mkdir(outDir)
    datField=allSim.xs(fieldname,level='fieldname')
    isimList=info[info['field']==fieldname]['num']
    print(fieldname)
    for isim in isimList:
        if not isim in allSim.index.levels[1]:
            continue
        outFname=os.path.join(outDir,'sim-%05d.fits' %isim)
        data=datField.xs(isim,level='isim').to_records(index=True)
        pyfits.writeto(outFname,data,overwrite=True)
```
