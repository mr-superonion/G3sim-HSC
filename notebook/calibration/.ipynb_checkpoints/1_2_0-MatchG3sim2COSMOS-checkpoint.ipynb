# Match the Sample 4 inputs to the 25.2 mag limit catalog

```python
%pylab inline
import os
import numpy as np
import astropy.io.fits as pyfits
import astropy.table as astTable
```

```python
cosmoname2= '/work/xiangchong.li/work/galsim_train/COSMOS_25.2_training_sample/real_galaxy_catalog_25.2_fits.fits' 

assert os.path.isfile(cosmoname2)
cosmocat2=astTable.Table.read(cosmoname2)
for pname in ['best','median','worst']:
    cosmoname=  '/work/xiangchong.li/work/galsim_train/parent_%s_processed/real_galaxy_catalog_%s.fits' %(pname,pname)
    assert os.path.isfile(cosmoname)
    cosmocat=astTable.Table.read(cosmoname)
    cosmocat['cosmos_index']=np.arange(len(cosmocat),dtype=int)
    
    cosmocat4=astTable.join(cosmocat,cosmocat2,'IDENT','left')
    cosmocat4=cosmocat4[np.argsort(cosmocat4['cosmos_index'])]
    cosmoname4=  '/work/xiangchong.li/work/galsim_train/real_galaxy_catalog_%s_fits.fits' %(pname)
    cosmocat4.write(cosmoname4,overwrite=True)
    # Make sure the index is the same as the origin catalog
    print(np.all(cosmocat['IDENT']==cosmocat4['IDENT']))
```

```python
z_cosmo=cosmocat4[~cosmocat4['zphot'].mask]['zphot']
print(np.percentile(z_cosmo,[0.,33.,66.,100.]))
plt.hist(z_cosmo,bins=64)
plt.xlabel('COSMOS z',fontsize=15)
plt.ylabel('prob',fontsize=15)
plt.tight_layout()
```

```python
ifile=0
pname=dict_parent[parent_idx[ifile]]
resname =   'wrkDir/pre/g3Calib/sample4-matched/src-%04d.fits'%(ifile)
prename =   '/work/xiangchong.li/work/S16ASimHSC/sample4/epoch_catalog-%04d-0.fits' %(ifile)
outname  =   'wrkDir/pre/g3Calib/sample4-COSMOS/src-%04d.fits' %(ifile)
cosmoname4=  '/work/xiangchong.li/work/galsim_train/real_galaxy_catalog_%s_fits.fits' %(pname)
assert os.path.isfile(resname)
assert os.path.isfile(prename)
assert os.path.isfile(cosmoname4)
```

```python
result=astTable.Table.read(resname)
precat=pyfits.getdata(prename)
cosmocat=pyfits.getdata(cosmoname4)
```

```python
preInd=np.int_(precat['x']//64)+np.int_(precat['y']//64)*100
preIndSort=np.argsort(preInd)
resInd=preIndSort[np.int_(result['base_SdssCentroid_x']//64)+np.int_(result['base_SdssCentroid_y']//64)*100]
cosmoInd=precat['cosmos_index'][resInd]
mag_cosmo=cosmocat['mag'][cosmoInd]
z_cosmo=cosmocat['zphot'][cosmoInd]
result['COSMOS_mag']=mag_cosmo
result['COSMOS_zphot']=z_cosmo
result.write(outname,overwrite=True)
```
